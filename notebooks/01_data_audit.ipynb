## Player Eligibility

For the performance analysis, players are included only if they have played at least 900 minutes.

This threshold is used to reduce the influence of very small playing-time samples while retaining a sufficiently large player pool for positional analysis.

- Total players in raw dataset: 517
- Eligible players: 301
- Minimum playing time: 900 minutes

```python
eligible_df = df[df["minutesPlayed"] >= 900].copy()
```
